# Problema: Aleta anular con resistencia de contacto

Una aleta anular de aluminio se instala alrededor de un tubo cilíndrico que transporta un fluido caliente. La pared exterior del tubo se mantiene a una temperatura uniforme de:

$$
T_w = 150\,^\circ\mathrm{C}
$$

El aire circundante está a:

$$
T_\infty = 25\,^\circ\mathrm{C}
$$

Entre el tubo y la aleta existe una resistencia térmica de contacto, cuya resistencia por unidad de área es:

$$
R_c'' = 3.0 \times 10^{-4}\ \mathrm{m^2\,K/W}
$$

$$t_e= 2 mm,  espacio entre aletas$$
La aleta posee las siguientes propiedades y dimensiones:

| Propiedad | Símbolo | Valor |
|---|---:|---:|
| Conductividad térmica | $k$ | $180\ \mathrm{W/(m\cdot K)}$ |
| Radio interior | $r_i$ | $10\ \mathrm{mm}$ |
| Radio exterior | $r_o$ | $40\ \mathrm{mm}$ |
| Espesor constante | $t$ | $1.5\ \mathrm{mm}$ |
| Coeficiente convectivo | $h$ | $45\ \mathrm{W/(m^2\cdot K)}$ |

Suponga régimen permanente, conducción unidimensional en dirección radial, propiedades constantes y transferencia por convección desde ambas caras de la aleta y desde su borde exterior. Desprecie los efectos de radiación.

La resistencia térmica de contacto total se expresa como:

$$
R_c = \frac{R_c''}{A_c}
$$

donde el área de contacto entre la aleta y el tubo es:

$$
A_c = 2\pi r_i t
$$

Defina el exceso de temperatura como:

$$
\theta(r) = T(r) - T_\infty
$$

La ecuación diferencial que gobierna la distribución de temperatura en una aleta anular es:

$$
\frac{d^2\theta}{dr^2}
+
\frac{1}{r}\frac{d\theta}{dr}
-
m^2\theta
=
0
$$

donde:

$$
m = \sqrt{\frac{2h}{kt}}
$$

Las condiciones de frontera son:

$$
T(r_i) = T_b
$$

$$
-k\left.\frac{d\theta}{dr}\right|_{r=r_o}
=
h\theta(r_o)
$$

Debido a la resistencia de contacto, la temperatura en la base de la aleta, $T_b$, es menor que la temperatura de la pared del tubo, $T_w$. El balance de energía en la interfaz tubo--aleta es:

$$
\dot{Q}_f
=
\frac{T_w - T_b}{R_c}
$$

El área convectiva total de la aleta se calcula mediante:

$$
A_f
=
2\pi\left(r_o^2-r_i^2\right)
+
2\pi r_o t
$$

La eficiencia de la aleta se define como:

$$
\eta_f
=
\frac{\dot{Q}_f}
{hA_f\left(T_b-T_\infty\right)}
$$

## Determine

1. La resistencia térmica de contacto total, $R_c$.

2. La temperatura de base de la aleta, $T_b$.

3. La tasa de transferencia de calor desde la aleta, $\dot{Q}_f$.

4. La eficiencia de la aleta, $\eta_f$.

5. La transferencia de calor en el caso ideal, suponiendo contacto perfecto:

$$
R_c = 0
$$

6. El porcentaje de reducción de transferencia de calor ocasionado por la resistencia de contacto:

$$
\%\,\text{reducción}
=
\frac{\dot{Q}_{f,\mathrm{ideal}}-\dot{Q}_{f,\mathrm{contacto}}}
{\dot{Q}_{f,\mathrm{ideal}}}
\times 100
$$

#calculo de la temperatura de la base
$$q_f=\frac{T_w-T_b}{R_c}=\frac{T_b-T_m}{R_f}  $$

```python
import numpy as np
from scipy.special import k1,k0,i1,i0

k=180
r1=10e-3
r2=40e-3
r2c=r2+0.5*t
t=1.5e-3
te=2e-3
h=45
Tw=150
Tm=25
Rcp=3e-4
Ac=2*np.pi*r1*t
Rc=Rcp/Ac
m=np.sqrt(2*h/(k*t))
nuf=2*r1/(m*(r2c**2-r1**2))*(k1(m*r1)*i1(m*r2c)-i1(m*r1)*k1(m*r2c))/(k0(m*r1)*i1(m*r2c)+i0(m*r1)*k1(m*r2c))
As=np.pi*(r2c**2-r1**2)*2
Rf=1/(h*As*nuf)
Req=Rc+Rf
qf=(Tw-Tm)/Req
Tb=Tw-qf*Rc
qe=h*(2*np.pi*r1*te)*(Tw-Tm)
qb=h*(2*np.pi*r1*t)*(Tw-Tm)
eff=qf/qb

print('qf=',qf)
print('nuf=',nuf)
print('Tb=',Tb)
print('qe=',qe)
print('efectividad=',eff)
```
